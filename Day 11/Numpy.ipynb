# Mathematical Ops on numpy array

```python
list1 = [1,2,3,4]
list2 = [3,4,5,6]
print(list1 + list2)
```

```python
a = np.random.randint(0,10,(3,3))
b = np.random.randint(10,20,(3,3))
print(a)
print(b)
```

# Array Manipulation

```python
a = np.random.randint(0,10,(2,3))
print(a)
print(a.shape)
```

```python
trans = np.transpose(a)
print(trans)
print(trans.shape)
```

```python
a = np.random.randint(0,10,(2,3))
print(a)
print(a.shape)
```

```python
trans2 = a.T
print(trans2)
print(trans2.shape)
```

```python
# reshaping a array
a = np.random.randint(0,10,(2,3))
print(a)
print(a.shape)
```

```python
b = a.reshape(3,2)
print(b)
print(b.shape)
```
